```python
# =========================================================
# INSTALLS
# =========================================================

!pip install -q ultralytics xgboost datasets

# =========================================================
# IMPORTS
# =========================================================

import os
import cv2
import torch
import random
import warnings
import numpy as np
import pandas as pd

from tqdm import tqdm
from ultralytics import YOLO
from xgboost import XGBClassifier
from torch.utils.data import DataLoader, Dataset
from transformers import (
    VideoMAEForVideoClassification,
    VideoMAEImageProcessor
)

warnings.filterwarnings("ignore")

# =========================================================
# DEVICE
# =========================================================

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)

# =========================================================
# CLASS NAMES
# =========================================================

class_names = [
    "cheating_copying",
    "cheating_gesture",
    "cheating_mobile",
    "cheating_notes",
    "normal"
]

label2id = {
    c:i for i,c in enumerate(class_names)
}

id2label = {
    i:c for c,i in label2id.items()
}

# =========================================================
# LOAD VIDEOMAE PROCESSOR
# =========================================================

processor = VideoMAEImageProcessor.from_pretrained(
    "MCG-NJU/videomae-small-finetuned-kinetics"
)

# =========================================================
# LOAD TRAINED VIDEOMAE MODEL
# =========================================================

model = VideoMAEForVideoClassification.from_pretrained(
    "MCG-NJU/videomae-small-finetuned-kinetics",
    num_labels=5,
    ignore_mismatched_sizes=True,
    label2id=label2id,
    id2label=id2label
)

# =========================================================
# LOAD YOUR TRAINED WEIGHTS
# =========================================================

model.load_state_dict(
    torch.load(
        "/content/drive/MyDrive/FYP/Models/best_videomae.pth",
        map_location=device
    )
)

model = model.to(device)

model.eval()

print("✅ VideoMAE Loaded")

# =========================================================
# LOAD YOLO MODELS
# =========================================================

mobile_model = YOLO(
    "/content/drive/MyDrive/FYP/Models/mobile_best.pt"
)

notes_model = YOLO(
    "/content/drive/MyDrive/FYP/Models/notes_best.pt"
)

print("✅ YOLO Models Loaded")

# =========================================================
# DUMMY DATASET AND DATALOADER FOR FEATURE EXTRACTION
# (REPLACE WITH YOUR ACTUAL DATA LOADING)
# =========================================================

class DummyVideoDataset(Dataset):
    def __init__(self, num_samples=100, clip_len=16, img_size=224, num_classes=5):
        self.num_samples = num_samples
        self.clip_len = clip_len
        self.img_size = img_size
        self.num_classes = num_classes

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        # Generate dummy video frames (e.g., random noise images)
        # In a real scenario, you would load actual video clips here
        video_frames = []
        for _ in range(self.clip_len):
            # Dummy frame: 3 channels (RGB), img_size x img_size
            frame = np.random.randint(0, 256, (self.img_size, self.img_size, 3), dtype=np.uint8)
            video_frames.append(frame)

        # Process frames using the VideoMAE processor
        inputs = processor(video_frames, return_tensors="pt")
        pixel_values = inputs["pixel_values"].squeeze(0) # Remove batch dimension added by processor

        # Generate a dummy label for classification (0 to num_classes-1)
        label = torch.tensor(random.randint(0, self.num_classes - 1), dtype=torch.long)

        return pixel_values, label

# Initialize dummy dataset and dataloader
# You should replace DummyVideoDataset with your actual dataset implementation
# and load your real video data here.
val_dataset = DummyVideoDataset(num_samples=100, clip_len=16, img_size=224, num_classes=5)
val_loader = DataLoader(val_dataset, batch_size=4)

print("✅ Dummy Dataset and DataLoader created (replace with actual data)")

# =========================================================
# FEATURE STORAGE
# =========================================================

fusion_features = []
fusion_labels = []

# =========================================================
# FEATURE EXTRACTION LOOP
# =========================================================

for pixel_values, labels in tqdm(val_loader):

    pixel_values = pixel_values.to(device)

    # =========================================
    # VIDEOMAE PREDICTIONS
    # =========================================

    with torch.no_grad():

        outputs = model(
            pixel_values=pixel_values
        )

        probs = torch.softmax(
            outputs.logits,
            dim=1
        ).cpu().numpy()

    batch_size = probs.shape[0]

    # =========================================
    # LOOP EACH SAMPLE
    # =========================================

    for i in range(batch_size):

        videomae_probs = probs[i]

        # =========================================
        # REAL YOLO FEATURES PLACEHOLDER
        # =========================================

        # Later replace with actual detections
        # for now safe random placeholders

        mobile_conf = np.random.uniform(0,1)

        notes_conf = np.random.uniform(0,1)

        # =========================================
        # FINAL FEATURE VECTOR
        # =========================================

        feature_vector = np.concatenate([

            videomae_probs,

            [mobile_conf],

            [notes_conf]

        ])

        fusion_features.append(
            feature_vector
        )

        fusion_labels.append(
            labels[i].item()
        )

# =========================================================
# CONVERT TO NUMPY
# =========================================================

fusion_features = np.array(
    fusion_features
)

fusion_labels = np.array(
    fusion_labels
)

print("✅ Feature Shape:", fusion_features.shape)
print("✅ Labels Shape:", fusion_labels.shape)

# =========================================================
# TRAIN XGBOOST FUSION MODEL
# =========================================================

fusion_model = XGBClassifier(

    n_estimators=200,
    max_depth=5,
    learning_rate=0.05,
    objective="multi:softprob",
    num_class=5,
    tree_method="hist",

    eval_metric="mlogloss"
)

fusion_model.fit(
    fusion_features,
    fusion_labels
)

print("✅ Fusion Model Trained")

# =========================================================
# SAVE FUSION MODEL
# =========================================================

import joblib

joblib.dump(
    fusion_model,
    "/content/drive/MyDrive/FYP/Models/fusion_xgboost.pkl"
)

print("✅ Fusion Model Saved")
```

```python
# =========================================================
# FINAL HYBRID EXAM CHEATING DETECTION PIPELINE
# STABLE VERSION
# VideoMAE + YOLO + YOLO Pose
# =========================================================

# =========================================================
# INSTALLS
# =========================================================

!pip install -q ultralytics transformers timm opencv-python

# =========================================================
# IMPORTS
# =========================================================

import cv2
import torch
import numpy as np
import pandas as pd
import time

from tqdm import tqdm
from ultralytics import YOLO

from transformers import (
    VideoMAEImageProcessor,
    VideoMAEForVideoClassification
)

from IPython.display import HTML
from base64 import b64encode

# =========================================================
# DEVICE
# =========================================================

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)

torch.backends.cudnn.benchmark = True

# =========================================================
# PATHS
# =========================================================

VIDEO_PATH = "/content/drive/MyDrive/OurDataset/dataset_f4/DSC_0129.MOV"

VIDEOMAE_MODEL_PATH = "/content/drive/MyDrive/FYP/Models/best_videomae.pth"

MOBILE_MODEL_PATH = "/content/drive/MyDrive/FYP/Models/mobile_best.pt"

NOTES_MODEL_PATH = "/content/drive/MyDrive/FYP/Models/notes_best.pt"

OUTPUT_VIDEO = "/content/drive/MyDrive/FYP/annotated/_0129_output.mp4"

OUTPUT_CSV = "/content/drive/MyDrive/FYP/annotated/pose_0129_log.csv"

# =========================================================
# CLASS NAMES
# =========================================================

class_names = [
    "copying",
    "gesture",
    "mobile",
    "notes",
    "normal"
]

# =========================================================
# LOAD MODELS
# =========================================================

tracker_model = YOLO("yolov8n.pt")

pose_model = YOLO("yolov8n-pose.pt")

mobile_model = YOLO(MOBILE_MODEL_PATH)

notes_model = YOLO(NOTES_MODEL_PATH)

# =========================================================
# LOAD VIDEOMAE
# =========================================================

processor = VideoMAEImageProcessor.from_pretrained(
    "MCG-NJU/videomae-small-finetuned-kinetics"
)

videomae_model = VideoMAEForVideoClassification.from_pretrained(
    "MCG-NJU/videomae-small-finetuned-kinetics",
    num_labels=5,
    ignore_mismatched_sizes=True
)

state_dict = torch.load(
    VIDEOMAE_MODEL_PATH,
    map_location=device
)

videomae_model.load_state_dict(
    state_dict,
    strict=False
)

videomae_model.to(device)

videomae_model.eval()

print("✅ All models loaded successfully")

# =========================================================
# SETTINGS
# =========================================================

CLIP_LEN = 16

IMG_SIZE = 224

FRAME_SKIP = 2

ACTION_INTERVAL = 12

YOLO_INTERVAL = 6

POSE_INTERVAL = 20

# =========================================================
# VIDEOMAE PREDICTION
# =========================================================

def predict_action(frames):

    frames = frames[-CLIP_LEN:]

    processed = []

    for frame in frames:

        frame = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        frame = cv2.resize(
            frame,
            (IMG_SIZE, IMG_SIZE)
        )

        processed.append(frame)

    inputs = processor(
        processed,
        return_tensors="pt"
    )

    pixel_values = inputs["pixel_values"].float().to(device)

    with torch.no_grad():

        outputs = videomae_model(
            pixel_values=pixel_values
        )

        probs = torch.softmax(
            outputs.logits,
            dim=1
        )

        pred = torch.argmax(
            probs,
            dim=1
        ).item()

        confidence = probs[0][pred].item()

    return (
        class_names[pred],
        confidence
    )

# =========================================================
# FAST YOLO DETECTION
# =========================================================

def detect_fast(model, image):

    results = model.predict(
        image,
        imgsz=320,
        conf=0.40,
        verbose=False,
        device=device
    )

    max_conf = 0.0

    if len(results[0].boxes) > 0:

        confs = results[0].boxes.conf.cpu().numpy()

        max_conf = float(np.max(confs))

    return max_conf

# =========================================================
# SMART YOLO POSE GESTURE DETECTION
# =========================================================

def detect_gesture_pose(crop):

    try:

        results = pose_model.predict(
            crop,
            imgsz=320,
            conf=0.35,
            verbose=False,
            device=device
        )

        if len(results[0].keypoints.xy) == 0:
            return False, 0.0

        kpts = results[0].keypoints.xy[0].cpu().numpy()

        nose = kpts[0]

        left_eye = kpts[1]
        right_eye = kpts[2]

        left_shoulder = kpts[5]
        right_shoulder = kpts[6]

        left_wrist = kpts[9]
        right_wrist = kpts[10]

        # =================================================
        # DISTANCES
        # =================================================

        left_face_dist = np.linalg.norm(
            left_wrist - nose
        )

        right_face_dist = np.linalg.norm(
            right_wrist - nose
        )

        # =================================================
        # STRONG CROSS BODY
        # =================================================

        left_cross = (
            left_wrist[0] >
            right_shoulder[0] + 40
        )

        right_cross = (
            right_wrist[0] <
            left_shoulder[0] - 40
        )

        # =================================================
        # BOTH HANDS CLOSE TO FACE
        # =================================================

        both_hands_face = (
            left_face_dist < 60 and
            right_face_dist < 60
        )

        # =================================================
        # HEAD SIDEWAYS
        # =================================================

        eye_distance = abs(
            left_eye[0] - right_eye[0]
        )

        head_sideways = eye_distance < 20

        suspicious = (

            left_cross or
            right_cross or
            both_hands_face or
            head_sideways
        )

        if suspicious:

            return True, 0.88

        return False, 0.0

    except:

        return False, 0.0

# =========================================================
# VIDEO
# =========================================================

cap = cv2.VideoCapture(VIDEO_PATH)

fps = cap.get(cv2.CAP_PROP_FPS)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))

height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print("FPS:", fps)

print("Frames:", total_frames)

# =========================================================
# OUTPUT VIDEO
# =========================================================

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    OUTPUT_VIDEO,
    fourcc,
    fps / FRAME_SKIP,
    (width, height)
)

# =========================================================
# VARIABLES
# =========================================================

student_buffers = {}

tracker_to_student = {}

student_predictions = {}

student_counter = 0

frame_no = 0

logs = []

# =========================================================
# TIMER
# =========================================================

start_time_total = time.time()

# =========================================================
# PROCESS VIDEO
# =========================================================

pbar = tqdm(total=total_frames)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_no += 1

    pbar.update(1)

    # =====================================================
    # FRAME SKIP
    # =====================================================

    if frame_no % FRAME_SKIP != 0:
        continue

    # =====================================================
    # PERSON TRACKING
    # =====================================================

    results = tracker_model.track(
        frame,
        persist=True,
        classes=[0],
        imgsz=640,
        conf=0.4,
        verbose=False,
        device=device
    )

    if results[0].boxes.id is not None:

        boxes = results[0].boxes.xyxy.cpu().numpy()

        ids = results[0].boxes.id.cpu().numpy().astype(int)

        for box, track_id in zip(boxes, ids):

            # =================================================
            # STUDENT IDS
            # =================================================

            if track_id not in tracker_to_student:

                student_counter += 1

                tracker_to_student[track_id] = student_counter

            sid = tracker_to_student[track_id]

            # =================================================
            # CROP
            # =================================================

            x1, y1, x2, y2 = map(int, box)

            x1 = max(0, x1)
            y1 = max(0, y1)

            x2 = min(width, x2)
            y2 = min(height, y2)

            crop = frame[y1:y2, x1:x2]

            if crop.size == 0:
                continue

            # =================================================
            # BUFFER
            # =================================================

            if sid not in student_buffers:

                student_buffers[sid] = []

            student_buffers[sid].append(crop)

            if len(student_buffers[sid]) > CLIP_LEN:

                student_buffers[sid] = student_buffers[sid][-CLIP_LEN:]

            # =================================================
            # DEFAULTS
            # =================================================

            label = "normal"

            confidence = 0.0

            mobile_conf = 0.0

            notes_conf = 0.0

            gesture_conf = 0.0

            gesture_found = False

            # =================================================
            # PREVIOUS PREDICTION
            # =================================================

            if sid in student_predictions:

                label = student_predictions[sid]["label"]

                confidence = student_predictions[sid]["confidence"]

            # =================================================
            # MAIN INFERENCE
            # =================================================

            if (
                len(student_buffers[sid]) >= CLIP_LEN
                and frame_no % ACTION_INTERVAL == 0
            ):

                # =============================================
                # VIDEOMAE
                # =============================================

                action_label, action_conf = predict_action(
                    student_buffers[sid]
                )

                label = action_label

                confidence = action_conf

                # =============================================
                # YOLO DETECTIONS
                # =============================================

                if frame_no % YOLO_INTERVAL == 0:

                    mobile_conf = detect_fast(
                        mobile_model,
                        crop
                    )

                    notes_conf = detect_fast(
                        notes_model,
                        crop
                    )

                # =============================================
                # POSE ONLY SOMETIMES
                # =============================================

                if frame_no % POSE_INTERVAL == 0:

                    gesture_found, gesture_conf = detect_gesture_pose(
                        crop
                    )

                # =============================================
                # FINAL FUSION
                # =============================================

                if mobile_conf > 0.70:

                    label = "mobile"

                    confidence = mobile_conf

                elif notes_conf > 0.70:

                    label = "notes"

                    confidence = notes_conf

                elif (
                    gesture_found and
                    action_label != "normal"
                ):

                    label = "gesture"

                    confidence = gesture_conf

                # =============================================
                # SAVE PREDICTION
                # =============================================

                student_predictions[sid] = {

                    "label": label,

                    "confidence": confidence
                }

                # =============================================
                # LOGS
                # =============================================

                logs.append({

                    "student_id": sid,

                    "time_sec": round(frame_no / fps, 2),

                    "action": label,

                    "confidence": round(confidence, 3),

                    "mobile_conf": round(mobile_conf, 3),

                    "notes_conf": round(notes_conf, 3),

                    "gesture_conf": round(gesture_conf, 3)
                })

            # =================================================
            # DRAW
            # =================================================

            color = (0,255,0)

            if label != "normal":

                color = (0,0,255)

            cv2.rectangle(
                frame,
                (x1,y1),
                (x2,y2),
                color,
                2
            )

            text = f"ID:{sid} {label} {confidence:.2f}"

            cv2.putText(
                frame,
                text,
                (x1, y1 - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (255,255,255),
                2
            )

    # =====================================================
    # WRITE OUTPUT
    # =====================================================

    out.write(frame)

# =========================================================
# RELEASE
# =========================================================

cap.release()

out.release()

pbar.close()

# =========================================================
# SAVE CSV
# =========================================================

log_df = pd.DataFrame(logs)

log_df.to_csv(
    OUTPUT_CSV,
    index=False
)

# =========================================================
# FINAL STATS
# =========================================================

total_time = time.time() - start_time_total

print("\n" + "="*60)
print("✅ PROCESSING COMPLETED")
print("="*60)

print(f"\n⏱ Total Time: {total_time:.2f} sec")

print(f"\n🚀 Effective FPS: {total_frames / total_time:.2f}")

print("\n📹 Annotated Video:")
print(OUTPUT_VIDEO)

print("\n📄 CSV Log:")
print(OUTPUT_CSV)

# =========================================================
# DISPLAY VIDEO
# =========================================================

mp4 = open(
    OUTPUT_VIDEO,
    "rb"
).read()

data_url = (
    "data:video/mp4;base64,"
    + b64encode(mp4).decode()
)

HTML(f"""
<video width=900 controls>
    <source src="{data_url}" type="video/mp4">
</video>
""")
```

```python
# =========================================================
# EXAM CHEATING DETECTION SYSTEM
# STABLE FINAL VERSION
# YOLO TRACKING + YOLO POSE + PHONE + NOTES
# =========================================================

# =========================================================
# INSTALL
# =========================================================

!pip install -q ultralytics opencv-python

# =========================================================
# IMPORTS
# =========================================================

import cv2
import numpy as np
from ultralytics import YOLO
from collections import defaultdict, deque

# =========================================================
# PATHS
# =========================================================

VIDEO_PATH = "/content/drive/MyDrive/ourDataset/dataset_f4/DSC_0129.MOV"

OUTPUT_PATH = "/content/drive/MyDrive/FYP/annotated/s1_0129_output.mp4"

NOTES_MODEL_PATH = "/content/drive/MyDrive/FYP/Models/notes_best.pt"

# =========================================================
# LOAD MODELS
# =========================================================

# PERSON + PHONE DETECTION
detector = YOLO("yolov8n.pt")

# POSE MODEL
pose_model = YOLO("yolov8n-pose.pt")

# CUSTOM NOTES MODEL
notes_model = YOLO(NOTES_MODEL_PATH)

print("✅ Models Loaded")

# =========================================================
# SETTINGS
# =========================================================

CLIP_MEMORY = 20

SCORE_THRESHOLD = 6

PHONE_CLASS_ID = 67

# =========================================================
# STORAGE
# =========================================================

history = defaultdict(lambda: deque(maxlen=CLIP_MEMORY))

scores = defaultdict(float)

tracker_map = {}

student_features = {}

streaks = defaultdict(lambda: {
    "glance":0,
    "lean":0,
    "reach":0
})

# =========================================================
# RE-ID FEATURE
# =========================================================

def extract_feature(crop):

    crop = cv2.resize(crop,(64,128))

    hist = cv2.calcHist(
        [crop],
        [0,1,2],
        None,
        [8,8,8],
        [0,256]*3
    )

    hist = cv2.normalize(hist,hist).flatten()

    return hist

# =========================================================
# MATCH STUDENT
# =========================================================

def match_student(feature):

    best_id = None

    best_score = 999

    for sid,feat in student_features.items():

        dist = np.linalg.norm(feature - feat)

        if dist < best_score:

            best_score = dist

            best_id = sid

    if best_score < 0.6:
        return best_id

    return None

# =========================================================
# RULES
# =========================================================

def side_glance(kp):

    nose = kp[0]
    leye = kp[1]
    reye = kp[2]

    lear = kp[3]
    rear = kp[4]

    if lear[2] < 0.3 or rear[2] < 0.3:
        return 1

    face_center = (leye[0] + reye[0]) / 2

    if abs(nose[0] - face_center) > 25:
        return 1

    return 0

# =========================================================

def leaning(kp):

    ls = kp[5]
    rs = kp[6]

    if abs(ls[1] - rs[1]) < 35:
        return 0

    return 1

# =========================================================

def reaching(kp):

    lw = kp[9]
    rw = kp[10]

    torso = (kp[5][:2] + kp[6][:2]) / 2

    if np.linalg.norm(lw[:2] - torso) > 150:
        return 1

    if np.linalg.norm(rw[:2] - torso) > 150:
        return 1

    return 0

# =========================================================

def hand_movement(prev, kp):

    if prev is None:
        return 0

    lw = kp[9]
    rw = kp[10]

    plw = prev[9]
    prw = prev[10]

    move = (
        np.linalg.norm(lw[:2]-plw[:2]) +
        np.linalg.norm(rw[:2]-prw[:2])
    )

    if move > 90:
        return 1

    return 0

# =========================================================
# DRAW POSE
# =========================================================

SKELETON = [
    (0,1),(0,2),(1,3),(2,4),
    (5,6),
    (5,9),(6,10),
    (5,11),(6,12),(11,12)
]

def draw_pose(frame, kp, offset):

    ox, oy = offset

    for i,j in SKELETON:

        if kp[i][2] > 0.3 and kp[j][2] > 0.3:

            x1,y1 = int(kp[i][0])+ox, int(kp[i][1])+oy
            x2,y2 = int(kp[j][0])+ox, int(kp[j][1])+oy

            cv2.line(frame,(x1,y1),(x2,y2),(255,0,0),2)

    for i in range(len(kp)):

        if i not in [7,8]:

            x,y,c = kp[i]

            if c > 0.3:

                cv2.circle(
                    frame,
                    (int(x)+ox,int(y)+oy),
                    4,
                    (0,255,255),
                    -1
                )

# =========================================================
# VIDEO
# =========================================================

cap = cv2.VideoCapture(VIDEO_PATH)

fps = cap.get(cv2.CAP_PROP_FPS)

if fps == 0:
    fps = 25

w = int(cap.get(3))
h = int(cap.get(4))

# =========================================================
# OUTPUT
# =========================================================

out = cv2.VideoWriter(
    OUTPUT_PATH,
    cv2.VideoWriter_fourcc(*'mp4v'),
    fps,
    (w,h)
)

# =========================================================
# MAIN
# =========================================================

sid_counter = 0

frame_no = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_no += 1

    # =====================================================
    # PERSON TRACKING
    # =====================================================

    results = detector.track(
        frame,
        persist=True,
        classes=[0],
        conf=0.4,
        verbose=False
    )

    # =====================================================
    # PHONE DETECTION (FULL FRAME)
    # =====================================================

    phone_results = detector.predict(
        frame,
        conf=0.35,
        verbose=False
    )

    phone_boxes = []

    for box in phone_results[0].boxes:

        cls = int(box.cls[0])

        if cls == PHONE_CLASS_ID:

            x1,y1,x2,y2 = map(
                int,
                box.xyxy[0]
            )

            phone_boxes.append(
                (x1,y1,x2,y2)
            )

            cv2.rectangle(
                frame,
                (x1,y1),
                (x2,y2),
                (0,0,255),
                2
            )

            cv2.putText(
                frame,
                "PHONE",
                (x1,y1-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0,0,255),
                2
            )

    # =====================================================
    # PERSON LOOP
    # =====================================================

    if results[0].boxes.id is not None:

        boxes = results[0].boxes.xyxy.cpu().numpy()

        ids = results[0].boxes.id.cpu().numpy().astype(int)

        for box,tid in zip(boxes,ids):

            x1,y1,x2,y2 = map(int,box)

            crop = frame[y1:y2,x1:x2]

            if crop.size == 0:
                continue

            # =================================================
            # STABLE ID
            # =================================================

            if tid in tracker_map:

                sid = tracker_map[tid]

            else:

                feature = extract_feature(crop)

                matched_id = match_student(feature)

                if matched_id is not None:

                    sid = matched_id

                else:

                    sid_counter += 1

                    sid = sid_counter

                    student_features[sid] = feature

                tracker_map[tid] = sid

            # =================================================
            # POSE
            # =================================================

            pose = pose_model(
                crop,
                verbose=False
            )

            if len(pose[0].keypoints.data) == 0:
                continue

            kp = pose[0].keypoints.data[0].cpu().numpy()

            prev = history[sid][-1] \
                if len(history[sid]) > 0 else None

            # =================================================
            # RULES
            # =================================================

            g = side_glance(kp)

            l = leaning(kp)

            r = reaching(kp)

            hm = hand_movement(prev, kp)

            streaks[sid]["glance"] = \
                streaks[sid]["glance"]+1 if g else 0

            streaks[sid]["lean"] = \
                streaks[sid]["lean"]+1 if l else 0

            streaks[sid]["reach"] = \
                streaks[sid]["reach"]+1 if r else 0

            score = 0

            if streaks[sid]["glance"] > 20:
                score += 2

            if streaks[sid]["lean"] > 25:
                score += 1

            if streaks[sid]["reach"] > 10:
                score += 2

            score += hm

            # =================================================
            # NOTES DETECTION
            # =================================================

            notes_results = notes_model.predict(
                crop,
                conf=0.50,
                verbose=False
            )

            notes_found = False

            if len(notes_results[0].boxes) > 0:

                notes_found = True

                score += 4

            # =================================================
            # PHONE INSIDE PERSON
            # =================================================

            phone_found = False

            for px1,py1,px2,py2 in phone_boxes:

                cx = (px1 + px2) // 2
                cy = (py1 + py2) // 2

                if (
                    x1 < cx < x2 and
                    y1 < cy < y2
                ):

                    phone_found = True

                    score += 5

            # =================================================
            # SCORE UPDATE
            # =================================================

            scores[sid] *= 0.85

            scores[sid] += score

            scores[sid] = min(scores[sid],10)

            history[sid].append(kp)

            # =================================================
            # STATUS
            # =================================================

            status = "NORMAL"

            color = (0,255,0)

            if phone_found:

                status = "MOBILE"

                color = (0,0,255)

            elif notes_found:

                status = "NOTES"

                color = (0,140,255)

            elif scores[sid] > SCORE_THRESHOLD:

                status = "SUSPICIOUS"

                color = (0,0,255)

            # =================================================
            # DRAW
            # =================================================

            cv2.rectangle(
                frame,
                (x1,y1),
                (x2,y2),
                color,
                3
            )

            cv2.putText(
                frame,
                f"ID:{sid} {status} {scores[sid]:.1f}",
                (x1,y1-10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                color,
                2
            )

            draw_pose(frame,kp,(x1,y1))

    # =====================================================
    # WRITE
    # =====================================================

    out.write(frame)

# =========================================================
# RELEASE
# =========================================================

cap.release()

out.release()

print("✅ FINAL OUTPUT SAVED")
print(OUTPUT_PATH)
```

```python
import cv2
import numpy as np

!pip install ultralytics
from tqdm import tqdm
from ultralytics import YOLO
from collections import defaultdict, deque


# Load models
tracker = YOLO("yolov8n.pt")        # YOLO for person detection/tracking
pose_model = YOLO("yolov8n-pose.pt")  # YOLO Pose for landmarks

# Video paths
VIDEO_PATH = "/content/drive/MyDrive/ourDataset/Testing samples/0122_g.MOV"
OUTPUT_PATH = "/content/drive/MyDrive/FYP/annotated/video_0122_g.mp4"

# Settings
SCORE_THRESHOLD = 6
CLIP_MEMORY = 20

# State
tracker_to_student = {}
student_id = 0
copy_scores = defaultdict(float)
history = defaultdict(lambda: deque(maxlen=CLIP_MEMORY))

# Rule functions (using COCO pose keypoints from YOLOv8 pose)
def side_glance(kp):
    nose, leye, reye, lear, rear = kp[0], kp[1], kp[2], kp[3], kp[4]
    # If one ear is not visible, or nose is far from face center, flag
    if lear[2] < 0.3 or rear[2] < 0.3:
        return 1
    face_center = (leye[0] + reye[0]) / 2
    return 1 if abs(nose[0] - face_center) > 25 else 0

def leaning(kp):
    ls, rs = kp[5], kp[6]  # left and right shoulders
    return 1 if abs(ls[1] - rs[1]) > 35 else 0

def reaching(kp):
    lw, rw = kp[9], kp[10]  # left and right wrists
    torso = (kp[5][:2] + kp[6][:2]) / 2
    if np.linalg.norm(lw[:2] - torso) > 150 or np.linalg.norm(rw[:2] - torso) > 150:
        return 1
    return 0

def hand_raise(kp):
    lw, rw = kp[9], kp[10]
    ls, rs = kp[5], kp[6]
    # Check if wrist is substantially above shoulder
    return (lw[1] < ls[1] - 20) or (rw[1] < rs[1] - 20)

# Draw pose skeleton and dashboard (optional)
def draw_pose(frame, kp, offset):
    ox, oy = offset
    SKELETON = [(0,1),(0,2),(1,3),(2,4),(5,6),(5,9),(6,10),(5,11),(6,12),(11,12)]
    for i,j in SKELETON:
        if kp[i][2]>0.3 and kp[j][2]>0.3:
            x1,y1 = int(kp[i][0])+ox, int(kp[i][1])+oy
            x2,y2 = int(kp[j][0])+ox, int(kp[j][1])+oy
            cv2.line(frame,(x1,y1),(x2,y2),(255,0,0),2)
    for i,(x,y,c) in enumerate(kp):
        if i not in [7,8] and c>0.3:
            cv2.circle(frame, (int(x)+ox,int(y)+oy), 4, (0,255,255), -1)

cap = cv2.VideoCapture(VIDEO_PATH)
fps = cap.get(cv2.CAP_PROP_FPS) or 25
width = int(cap.get(3)); height = int(cap.get(4))
out = cv2.VideoWriter(OUTPUT_PATH, cv2.VideoWriter_fourcc(*'mp4v'), fps, (width, height))

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Detect and track persons
    results = tracker.track(frame, persist=True, classes=[0], conf=0.4, verbose=False)

    # Check if there are any tracked person IDs before proceeding
    if results[0].boxes.id is None or len(results[0].boxes.id) == 0:
        out.write(frame)
        continue

    boxes = results[0].boxes.xyxy.cpu().numpy()
    ids = results[0].boxes.id.cpu().numpy().astype(int)

    for box, tid in zip(boxes, ids):
        x1,y1,x2,y2 = map(int, box)
        crop = frame[y1:y2, x1:x2]
        if crop.size == 0:
            continue

        # Assign a stable student ID
        if tid not in tracker_to_student:
            student_id += 1
            tracker_to_student[tid] = student_id
        sid = tracker_to_student[tid]

        # Pose landmarks
        pose_res = pose_model(crop, verbose=False)
        if len(pose_res[0].keypoints.data) == 0:
            continue
        kp = pose_res[0].keypoints.data[0].cpu().numpy()
        prev_kp = history[sid][-1] if history[sid] else None

        # Update copying scores
        g = side_glance(kp); l = leaning(kp); r = reaching(kp)
        if g:  copy_scores[sid] += 2
        if l:  copy_scores[sid] += 1
        if r:  copy_scores[sid] += 2
        # Cap score and apply decay
        copy_scores[sid] = min(copy_scores[sid], 10)
        if not (g or l or r):
            copy_scores[sid] *= 0.85

        history[sid].append(kp)

        # Detect gesture (raised hand)
        gesture = False
        if hand_raise(kp):
            # Require persistence: check previous frame too if available
            if prev_kp is None or hand_raise(prev_kp):
                gesture = True

        # Determine label and color
        label = "normal"; conf = 0.0; color = (0,255,0)
        if gesture:
            label = "cheating_gesture"; conf = 0.90; color = (0,0,255)
        elif copy_scores[sid] > SCORE_THRESHOLD:
            label = "cheating_copying"
            conf = min(copy_scores[sid]/10, 0.95); color = (0,165,255)

        # Draw box and label
        cv2.rectangle(frame, (x1,y1),(x2,y2), color, 3)
        cv2.putText(frame, f"ID:{sid} {label} {conf:.2f}", (x1,y1-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, color, 2)
        draw_pose(frame, kp, (x1,y1))

    out.write(frame)

cap.release()
out.release()
print("✅ Processing complete. Output saved to", OUTPUT_PATH)
```

# fusion with VideoMAE

```python
import cv2
import torch
import numpy as np
import pandas as pd

from ultralytics import YOLO
from collections import defaultdict, deque

from transformers import (
    VideoMAEImageProcessor,
    VideoMAEForVideoClassification
)

# =========================================================
# DEVICE
# =========================================================

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)

# =========================================================
# PATHS
# =========================================================

VIDEO_PATH = "/content/drive/MyDrive/ourDataset/Testing samples/0129(c).MOV"

VIDEOMAE_MODEL = "/content/drive/MyDrive/FYP/Models/final_model.pth"

OUTPUT_VIDEO = "/content/drive/MyDrive/FYP/annotated/final_output.mp4"

CSV_OUTPUT = "/content/drive/MyDrive/FYP/annotated/final_logs.csv"

# =========================================================
# CLASSES
# =========================================================

class_names = [
    "cheating_copying",
    "cheating_gesture",
    "normal"
]

# =========================================================
# MODELS
# =========================================================

tracker = YOLO("yolov8n.pt")

pose_model = YOLO("yolov8n-pose.pt")

# =========================================================
# VIDEOMAE
# =========================================================

processor = VideoMAEImageProcessor.from_pretrained(
    "MCG-NJU/videomae-base-finetuned-kinetics"
)

videomae = VideoMAEForVideoClassification.from_pretrained(
    "MCG-NJU/videomae-base-finetuned-kinetics",
    num_labels=3,
    ignore_mismatched_sizes=True
)

checkpoint = torch.load(
    VIDEOMAE_MODEL,
    map_location=device
)

if "model_state_dict" in checkpoint:
    checkpoint = checkpoint["model_state_dict"]

model_dict = videomae.state_dict()

filtered = {}

for k, v in checkpoint.items():

    if k in model_dict:

        if v.shape == model_dict[k].shape:

            filtered[k] = v

videomae.load_state_dict(
    filtered,
    strict=False
)

videomae.to(device)

videomae.eval()

print("✅ VideoMAE Loaded")

# =========================================================
# SETTINGS
# =========================================================

CLIP_LEN = 16

FRAME_SKIP = 2

# STRICT TEMPORAL THRESHOLDS

COPY_TEMPORAL_FRAMES = 8
GESTURE_TEMPORAL_FRAMES = 7

# =========================================================
# STORAGE
# =========================================================

student_buffers = defaultdict(list)

prediction_history = defaultdict(
    lambda: deque(maxlen=15)
)

tracker_map = {}

sid_counter = 0

# temporal counters

side_counter = defaultdict(int)

gesture_counter = defaultdict(int)

copy_counter = defaultdict(int)

# =========================================================
# VIDEOMAE PREDICTION
# =========================================================

def predict_videomae(frames):

    processed = []

    for f in frames[-CLIP_LEN:]:

        rgb = cv2.cvtColor(
            f,
            cv2.COLOR_BGR2RGB
        )

        rgb = cv2.resize(
            rgb,
            (224,224)
        )

        processed.append(rgb)

    inputs = processor(
        processed,
        return_tensors="pt"
    )

    pixel_values = inputs[
        "pixel_values"
    ].to(device)

    with torch.no_grad():

        outputs = videomae(
            pixel_values=pixel_values
        )

        probs = torch.softmax(
            outputs.logits,
            dim=1
        )

        pred = torch.argmax(
            probs,
            dim=1
        ).item()

        conf = probs[0][pred].item()

    return class_names[pred], conf

# =========================================================
# LANDMARK CONNECTIONS
# =========================================================

SKELETON = [

    (0,1),(0,2),
    (1,3),(2,4),

    (5,6),

    (5,7),(7,9),
    (6,8),(8,10),

    (5,11),(6,12),

    (11,12),

    (11,13),(13,15),
    (12,14),(14,16)
]

# =========================================================
# RULES
# =========================================================

def side_glance(kp):

    nose = kp[0]

    leye = kp[1]

    reye = kp[2]

    if leye[2] < 0.4 or reye[2] < 0.4:
        return False

    eye_center = (
        leye[0] + reye[0]
    ) / 2

    dist = abs(
        nose[0] - eye_center
    )

    return dist > 15

# =========================================================

def leaning(kp):

    ls = kp[5]

    rs = kp[6]

    diff = abs(
        ls[1] - rs[1]
    )

    return diff > 35

# =========================================================

def looking_down(kp):

    nose = kp[0]

    ls = kp[5]

    rs = kp[6]

    shoulder_y = (
        ls[1] + rs[1]
    ) / 2

    return nose[1] > shoulder_y - 20

# =========================================================
# STRICT GESTURE RULE
# =========================================================

def cheating_gesture(kp):

    lw = kp[9]
    rw = kp[10]

    ls = kp[5]
    rs = kp[6]

    le = kp[7]
    re = kp[8]

    # hand above shoulder

    left_up = lw[1] < ls[1] + 10
    right_up = rw[1] < rs[1] + 10

    # hand NOT near writing position

    left_far = abs(
        lw[0] - ls[0]
    ) > 35

    right_far = abs(
        rw[0] - rs[0]
    ) > 35

    # elbow bent

    left_arm = np.linalg.norm(
        lw[:2] - le[:2]
    )

    right_arm = np.linalg.norm(
        rw[:2] - re[:2]
    )

    left_valid = (
        left_up and
        left_far and
        left_arm > 40
    )

    right_valid = (
        right_up and
        right_far and
        right_arm > 40
    )

    return left_valid or right_valid

# =========================================================
# LOOKING OTHER STUDENT
# =========================================================

def looking_other_student(kp1, kp2):

    nose1 = kp1[0]

    eyes_center = (
        kp1[1][:2] +
        kp1[2][:2]
    ) / 2

    direction = nose1[:2] - eyes_center

    target = kp2[0][:2] - nose1[:2]

    dot = np.dot(
        direction,
        target
    )

    return dot > 0

# =========================================================
# VIDEO
# =========================================================

cap = cv2.VideoCapture(VIDEO_PATH)

fps = cap.get(cv2.CAP_PROP_FPS)

width = int(cap.get(3))

height = int(cap.get(4))

out = cv2.VideoWriter(
    OUTPUT_VIDEO,
    cv2.VideoWriter_fourcc(*'mp4v'),
    fps / FRAME_SKIP,
    (width,height)
)

# =========================================================
# LOGS
# =========================================================

logs = []

frame_no = 0

# =========================================================
# MAIN LOOP
# =========================================================

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_no += 1

    if frame_no % FRAME_SKIP != 0:
        continue

    # =====================================================
    # DETECTION + TRACKING
    # =====================================================

    results = tracker.track(
        frame,
        persist=True,
        classes=[0],
        conf=0.45,
        verbose=False
    )

    if results[0].boxes.id is None:

        out.write(frame)

        continue

    boxes = results[0].boxes.xyxy.cpu().numpy()

    ids = results[0].boxes.id.cpu().numpy().astype(int)

    students = []

    # =====================================================
    # FIRST PASS
    # =====================================================

    for box, tid in zip(boxes, ids):

        x1,y1,x2,y2 = map(int, box)

        crop = frame[y1:y2, x1:x2]

        if crop.size == 0:
            continue

        if tid not in tracker_map:

            sid_counter += 1

            tracker_map[tid] = sid_counter

        sid = tracker_map[tid]

        # =================================================
        # BUFFER
        # =================================================

        student_buffers[sid].append(crop)

        if len(student_buffers[sid]) > CLIP_LEN:

            student_buffers[sid] = \
            student_buffers[sid][-CLIP_LEN:]

        # =================================================
        # POSE
        # =================================================

        pose = pose_model(
            crop,
            verbose=False
        )

        if len(
            pose[0].keypoints.data
        ) == 0:

            continue

        kp = pose[0].keypoints.data[
            0
        ].cpu().numpy()

        students.append({

            "sid": sid,
            "box": (x1,y1,x2,y2),
            "kp": kp,
            "crop": crop
        })

    # =====================================================
    # SECOND PASS
    # =====================================================

    for student in students:

        sid = student["sid"]

        x1,y1,x2,y2 = student["box"]

        kp = student["kp"]

        crop = student["crop"]

        # =================================================
        # RULES
        # =================================================

        side = side_glance(kp)

        lean = leaning(kp)

        gesture = cheating_gesture(kp)

        look_other = False

        for other in students:

            if other["sid"] == sid:
                continue

            if looking_other_student(
                kp,
                other["kp"]
            ):

                look_other = True
                break

        # =================================================
        # TEMPORAL COPYING
        # =================================================

        if side and look_other:

            side_counter[sid] += 1

        else:

            side_counter[sid] = max(
                0,
                side_counter[sid] - 1
            )

        # =================================================
        # STRICT COPYING
        # =================================================

        copying = (
            side_counter[sid]
            >= COPY_TEMPORAL_FRAMES
        )

        # =================================================
        # STRICT GESTURE
        # =================================================

        if gesture:

            gesture_counter[sid] += 1

        else:

            gesture_counter[sid] = max(
                0,
                gesture_counter[sid] - 1
            )

        gesture_detected = (
            gesture_counter[sid]
            >= GESTURE_TEMPORAL_FRAMES
        )

        # =================================================
        # VIDEOMAE
        # =================================================

        videomae_label = "normal"

        videomae_conf = 0.5

        if len(
            student_buffers[sid]
        ) >= CLIP_LEN:

            videomae_label, videomae_conf = \
            predict_videomae(
                student_buffers[sid]
            )

        # =================================================
        # FINAL FUSION
        # =================================================

        final_label = "normal"

        final_conf = 0.70

        # STRICT RULES FIRST

        if gesture_detected:

            final_label = "cheating_gesture"

            final_conf = max(
                0.80,
                videomae_conf
            )

        elif copying:

            final_label = "cheating_copying"

            final_conf = max(
                0.80,
                videomae_conf
            )

        # MODEL SUPPORT

        elif (
            videomae_label != "normal"
            and videomae_conf > 0.85
        ):

            final_label = videomae_label

            final_conf = videomae_conf

        # =================================================
        # TEMPORAL SMOOTHING
        # =================================================

        prediction_history[sid].append(
            final_label
        )

        stable_label = max(
            set(prediction_history[sid]),
            key=prediction_history[sid].count
        )

        # =================================================
        # DRAW SKELETON
        # =================================================

        for p1, p2 in SKELETON:

            if (
                kp[p1][2] > 0.3
                and
                kp[p2][2] > 0.3
            ):

                xA, yA = int(kp[p1][0]), int(kp[p1][1])

                xB, yB = int(kp[p2][0]), int(kp[p2][1])

                cv2.line(
                    crop,
                    (xA,yA),
                    (xB,yB),
                    (0,255,255),
                    2
                )

        # =================================================
        # DRAW POINTS
        # =================================================

        for point in kp:

            px, py, conf = point

            if conf > 0.3:

                cv2.circle(
                    crop,
                    (int(px), int(py)),
                    4,
                    (255,0,0),
                    -1
                )

        # =================================================
        # DRAW BOX
        # =================================================

        color = (0,255,0)

        if stable_label != "normal":

            color = (0,0,255)

        cv2.rectangle(
            frame,
            (x1,y1),
            (x2,y2),
            color,
            3
        )

        # =================================================
        # TEXT
        # =================================================

        text = (
            f"ID:{sid} "
            f"{stable_label} "
            f"{final_conf:.2f}"
        )

        cv2.putText(
            frame,
            text,
            (x1,y1-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            color,
            2
        )

        # =================================================
        # DEBUG
        # =================================================

        debug = (
            f"S:{side_counter[sid]} "
            f"G:{gesture_counter[sid]}"
        )

        cv2.putText(
            frame,
            debug,
            (x1,y2+20),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255,255,0),
            2
        )

        # =================================================
        # LOGS
        # =================================================

        logs.append({

            "frame": frame_no,

            "student_id": sid,

            "videomae_label": videomae_label,

            "videomae_conf": videomae_conf,

            "copy_counter": side_counter[sid],

            "gesture_counter": gesture_counter[sid],

            "final_label": stable_label,

            "final_conf": final_conf
        })

    out.write(frame)

# =========================================================
# SAVE
# =========================================================

cap.release()

out.release()

df = pd.DataFrame(logs)

df.to_csv(
    CSV_OUTPUT,
    index=False
)

print("\n✅ DONE")

print("Video Saved:", OUTPUT_VIDEO)

print("CSV Saved:", CSV_OUTPUT)
```

```python
# =========================================================
# EXAM CHEATING DETECTION
# MOBILE PHONE USAGE DETECTION
# READY TO RUN - GOOGLE COLAB
# =========================================================

# =========================================================
# INSTALL
# =========================================================

!pip install -q ultralytics mediapipe opencv-python opencv-contrib-python pandas

print("✅ INSTALLATION FINISHED")

# =========================================================
# IMPORTS
# =========================================================

import cv2
import mediapipe as mp
import numpy as np
import pandas as pd
import torch

from ultralytics import YOLO
from collections import defaultdict, deque

# =========================================================
# VERIFY
# =========================================================

print("===================================")
print("OpenCV:", cv2.__version__)
print("MediaPipe:", mp.__version__)
print("NumPy:", np.__version__)
print("CUDA Available:", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))

print("===================================")

# =========================================================
# MEDIAPIPE
# =========================================================

mp_pose = mp.solutions.pose
mp_hands = mp.solutions.hands
mp_face_mesh = mp.solutions.face_mesh
mp_draw = mp.solutions.drawing_utils

pose = mp_pose.Pose(
    static_image_mode=False,
    model_complexity=1,
    smooth_landmarks=True,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)

hands = mp_hands.Hands(
    static_image_mode=False,
    max_num_hands=2,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)

face_mesh = mp_face_mesh.FaceMesh(
    static_image_mode=False,
    max_num_faces=1,
    refine_landmarks=True,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)

print("✅ MediaPipe Loaded Successfully")

# =========================================================
# PATHS
# =========================================================

VIDEO_PATH = "/content/drive/MyDrive/ourDataset/Testing samples/0136(m).MOV"

OUTPUT_VIDEO = "/content/drive/MyDrive/FYP/annotated/136_mobile_cheating_output.mp4"

CSV_OUTPUT = "/content/drive/MyDrive/FYP/annotated/136_mobile_cheating_logs.csv"

# =========================================================
# LOAD YOLO
# =========================================================

model = YOLO("yolov8m.pt")

# =========================================================
# SETTINGS
# =========================================================

PHONE_CLASS_ID = 67
PERSON_CLASS_ID = 0

FRAME_SKIP = 2

HAND_PHONE_IOU_THRESHOLD = 0.05

CHEATING_SCORE_THRESHOLD = 8

# =========================================================
# STORAGE
# =========================================================

tracker_map = {}

sid_counter = 0

cheating_scores = defaultdict(float)

prediction_history = defaultdict(
    lambda: deque(maxlen=20)
)

logs = []

# =========================================================
# IOU FUNCTION
# =========================================================

def calculate_iou(boxA, boxB):

    xA = max(boxA[0], boxB[0])
    yA = max(boxA[1], boxB[1])

    xB = min(boxA[2], boxB[2])
    yB = min(boxA[3], boxB[3])

    interW = max(0, xB - xA)
    interH = max(0, yB - yA)

    interArea = interW * interH

    if interArea == 0:
        return 0

    boxAArea = (
        (boxA[2] - boxA[0]) *
        (boxA[3] - boxA[1])
    )

    boxBArea = (
        (boxB[2] - boxB[0]) *
        (boxB[3] - boxB[1])
    )

    return interArea / float(
        boxAArea + boxBArea - interArea
    )

# =========================================================
# FINGER OPEN DETECTION
# =========================================================

def count_open_fingers(hand_landmarks):

    count = 0

    lm = hand_landmarks.landmark

    # thumb
    if lm[4].x < lm[3].x:
        count += 1

    # fingers
    for tip in [8, 12, 16, 20]:

        if lm[tip].y < lm[tip - 2].y:
            count += 1

    return count

# =========================================================
# HAND BOUNDING BOX
# =========================================================

def get_hand_box(hand_landmarks, w, h):

    xs = []
    ys = []

    for lm in hand_landmarks.landmark:

        xs.append(int(lm.x * w))
        ys.append(int(lm.y * h))

    return [
        min(xs),
        min(ys),
        max(xs),
        max(ys)
    ]

# =========================================================
# VIDEO
# =========================================================

cap = cv2.VideoCapture(VIDEO_PATH)

if not cap.isOpened():
    raise Exception(f"Cannot open video: {VIDEO_PATH}")

fps = cap.get(cv2.CAP_PROP_FPS)

if fps == 0:
    fps = 30

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

out = cv2.VideoWriter(
    OUTPUT_VIDEO,
    cv2.VideoWriter_fourcc(*'mp4v'),
    max(fps / FRAME_SKIP, 1),
    (width, height)
)

frame_no = 0

print("\n✅ Processing Started...\n")

# =========================================================
# MAIN LOOP
# =========================================================

while True:

    ret, frame = cap.read()

    if not ret:
        break

    frame_no += 1

    if frame_no % FRAME_SKIP != 0:
        continue

    # =====================================================
    # YOLO TRACKING
    # =====================================================

    results = model.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        verbose=False,
        conf=0.35,
        iou=0.5
    )

    if (
        results is None or
        len(results) == 0 or
        results[0].boxes is None or
        results[0].boxes.id is None
    ):
        out.write(frame)
        continue

    boxes = results[0].boxes.xyxy.cpu().numpy()

    classes = results[0].boxes.cls.cpu().numpy().astype(int)

    ids = results[0].boxes.id.cpu().numpy().astype(int)

    # =====================================================
    # DETECT PHONES
    # =====================================================

    phone_boxes = []

    for box, cls in zip(boxes, classes):

        if cls == PHONE_CLASS_ID:

            x1, y1, x2, y2 = map(int, box)

            phone_boxes.append([x1, y1, x2, y2])

            cv2.rectangle(
                frame,
                (x1, y1),
                (x2, y2),
                (0, 165, 255),
                3
            )

            cv2.putText(
                frame,
                "PHONE",
                (x1, y1 - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 165, 255),
                2
            )

    # =====================================================
    # PERSON LOOP
    # =====================================================

    for box, cls, tid in zip(boxes, classes, ids):

        if cls != PERSON_CLASS_ID:
            continue

        x1, y1, x2, y2 = map(int, box)

        x1 = max(0, x1)
        y1 = max(0, y1)

        x2 = min(width, x2)
        y2 = min(height, y2)

        crop = frame[y1:y2, x1:x2].copy()

        if crop.size == 0:
            continue

        # =================================================
        # STABLE IDS
        # =================================================

        if tid not in tracker_map:

            sid_counter += 1

            tracker_map[tid] = sid_counter

        sid = tracker_map[tid]

        cheating = False

        cheating_reason = []

        # =================================================
        # MEDIAPIPE
        # =================================================

        crop_rgb = cv2.cvtColor(
            crop,
            cv2.COLOR_BGR2RGB
        )

        pose_results = pose.process(crop_rgb)

        hand_results = hands.process(crop_rgb)

        face_results = face_mesh.process(crop_rgb)

        # =================================================
        # DRAW POSE
        # =================================================

        if pose_results.pose_landmarks:

            mp_draw.draw_landmarks(
                crop,
                pose_results.pose_landmarks,
                mp_pose.POSE_CONNECTIONS
            )

        # =================================================
        # DRAW FACE
        # =================================================

        if face_results.multi_face_landmarks:

            for face_landmarks in face_results.multi_face_landmarks:

                mp_draw.draw_landmarks(
                    crop,
                    face_landmarks,
                    mp_face_mesh.FACEMESH_TESSELATION,
                    landmark_drawing_spec=None,
                    connection_drawing_spec=
                    mp_draw.DrawingSpec(
                        color=(255,255,0),
                        thickness=1
                    )
                )

        # =================================================
        # HAND ANALYSIS
        # =================================================

        hand_phone_overlap = False

        finger_signal = 0

        if hand_results.multi_hand_landmarks:

            for hand_landmarks in hand_results.multi_hand_landmarks:

                mp_draw.draw_landmarks(
                    crop,
                    hand_landmarks,
                    mp_hands.HAND_CONNECTIONS
                )

                hand_box = get_hand_box(
                    hand_landmarks,
                    crop.shape[1],
                    crop.shape[0]
                )

                global_hand_box = [

                    hand_box[0] + x1,
                    hand_box[1] + y1,
                    hand_box[2] + x1,
                    hand_box[3] + y1
                ]

                open_fingers = count_open_fingers(
                    hand_landmarks
                )

                finger_signal = max(
                    finger_signal,
                    open_fingers
                )

                for pbox in phone_boxes:

                    iou = calculate_iou(
                        global_hand_box,
                        pbox
                    )

                    if iou > HAND_PHONE_IOU_THRESHOLD:

                        hand_phone_overlap = True
                        break

        # =================================================
        # RULE 1
        # =================================================

        if hand_phone_overlap:

            cheating_scores[sid] += 4

            cheating_reason.append(
                "HandPhoneOverlap"
            )

        # =================================================
        # RULE 2
        # Finger signal only counts with phone overlap
        # =================================================

        if hand_phone_overlap and finger_signal >= 3:

            cheating_scores[sid] += 2

            cheating_reason.append(
                f"FingerSignal:{finger_signal}"
            )

        # =================================================
        # DECAY
        # =================================================

        cheating_scores[sid] *= 0.92

        cheating_scores[sid] = min(
            cheating_scores[sid],
            10
        )

        # =================================================
        # FINAL DECISION
        # =================================================

        confidence = min(
            cheating_scores[sid] / 10,
            1.0
        )

        if cheating_scores[sid] >= CHEATING_SCORE_THRESHOLD:

            cheating = True

        # =================================================
        # TEMPORAL SMOOTHING
        # =================================================

        prediction_history[sid].append(
            cheating
        )

        stable_prediction = sum(
            prediction_history[sid]
        ) > (
            len(prediction_history[sid]) * 0.5
        )

        # =================================================
        # PUT CROP BACK
        # =================================================

        frame[y1:y2, x1:x2] = crop

        # =================================================
        # DRAW
        # =================================================

        color = (0,255,0)

        label = "NORMAL"

        if stable_prediction:

            color = (0,0,255)

            label = f"cheating_mobile_use {confidence:.2f}"

        cv2.rectangle(
            frame,
            (x1,y1),
            (x2,y2),
            color,
            3
        )

        cv2.putText(
            frame,
            f"ID:{sid} {label}",
            (x1,y1-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            color,
            2
        )

        debug_text = f"Score:{cheating_scores[sid]:.1f}"

        cv2.putText(
            frame,
            debug_text,
            (x1,y2+20),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255,255,0),
            2
        )

        # =================================================
        # SAVE LOG
        # =================================================

        logs.append({

            "frame": frame_no,

            "student_id": sid,

            "cheating_score":
            cheating_scores[sid],

            "finger_signal":
            finger_signal,

            "hand_phone_overlap":
            hand_phone_overlap,

            "final_prediction":
            label,

            "confidence":
            confidence
        })

    # =====================================================
    # WRITE FRAME
    # =====================================================

    out.write(frame)

# =========================================================
# SAVE
# =========================================================

cap.release()

out.release()

cv2.destroyAllWindows()

df = pd.DataFrame(logs)

df.to_csv(
    CSV_OUTPUT,
    index=False
)

print("\n===================================")
print("✅ DONE")
print("===================================")
print("Video Saved:", OUTPUT_VIDEO)
print("CSV Saved:", CSV_OUTPUT)
print("===================================")
```
